## 2Q)

Generate a list of dates from '2023-01-01' to '2023-01-15' using a recursive CTE.


## Q1 Solution
<pre>
mysql> CREATE TABLE orders (
    ->     order_id int,
    ->     invoice_date date
    -> );
Query OK, 0 rows affected (0.03 sec)

mysql> INSERT INTO orders (order_id,invoice_date) VALUES
    -> (330115,'01/03/24'),
    -> (330120,'01/03/24'),
    -> (330121,'01/03/24'),
    -> (330122,'02/03/24'),
    -> (330125,'02/03/24');
Query OK, 5 rows affected, 5 warnings (0.01 sec)
Records: 5  Duplicates: 0  Warnings: 5


mysql> WITH RECURSIVE all_ids as (
    ->     SELECT MIN(order_id) as order_id
    ->     FROM orders
    ->     UNION ALL
    ->     SELECT order_id + 1
    ->     FROM all_ids
    ->     WHERE order_id < (SELECT MAX(order_id) FROM orders)
    -> )
    -> SELECT a.order_id
    -> FROM all_ids a
    -> LEFT JOIN orders o ON a.order_id = o.order_id
    -> WHERE o.order_id IS NULL
    -> ORDER BY a.order_id;

## Q2 Solution
<pre>

mysql> WITH RECURSIVE DateSeries AS (
    ->     -- Anchor member: Defines the starting point of the series
    ->     SELECT CAST('2023-01-01' AS DATE) AS CalendarDate
    ->     UNION ALL
    ->     -- Recursive member: Generates subsequent dates
    ->     SELECT DATE_ADD(CalendarDate, INTERVAL 1 DAY)
    ->     FROM DateSeries
    ->     WHERE CalendarDate < '2023-01-15'
    -> )
    -> SELECT CalendarDate
    -> FROM DateSeries
    -> ORDER BY CalendarDate;
