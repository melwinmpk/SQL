Write a query to output the start and end dates of projects listed by the number of days it took to complete the project in ascending order. If there is more than one project that have the same number of completion days, then order by the start date of the project.

Correct Answer

```python
SELECT A.Start_Date, MIN(B.End_Date) 
FROM 
(Select Start_Date, End_Date FROM Projects WHERE Start_Date Not IN (SELECT End_Date FROM Projects)) A
INNER JOIN 
(Select Start_Date, End_Date FROM Projects WHERE End_Date Not in (SELECT Start_Date FROM Projects)) B
 WHERE A.Start_Date < B.End_Date
 GROUP BY A.Start_Date
 ORDER BY MIN(B.End_Date) - A.Start_Date ASC, A.Start_Date ASC
 ;
```
